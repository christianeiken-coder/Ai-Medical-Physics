```python
from torch.utils.data import Dataset
from torch.utils.data import DataLoader, WeightedRandomSampler
import torchvision.transforms.functional as ttf
import pandas as pd
import torchvision.transforms as T
import torch
from PIL import Image
import numpy as np


class LiTS_Dataset(Dataset):

    def __init__(self, csv: str, mode: str, transform=None):

        self.csv = csv
        self.transform = transform
        self.data = pd.read_csv(self.csv)
        self.mode = mode
        assert mode in ["train", "val", "test"]

    def __len__(self):

        return len(self.data)

    def __getitem__(self, idx):

        file = self.data.loc[idx, "filename"]
        with Image.open(f"./Clean_LiTS/{self.mode}/{file}") as f:
            f = f.convert("L")
            image = ttf.pil_to_tensor(f)   # [1,H,W]

        image = image.to(dtype=torch.float32)

        mean = image.mean()
        std = image.std()

        if std > 0:
          image = (image - mean) / std
        else:
          image = image - mean



        _, H, W = image.shape
        image = image[:, :, W // 2:]

        liver_visible  = int(self.data.loc[idx, "liver_visible"])
        lesion_visible = int(self.data.loc[idx, "lesion_visible"])

        # multi-label: [liver, tumor]
        target = torch.tensor([liver_visible, lesion_visible],
                              dtype=torch.float32)

        if self.transform:
            image = self.transform(image)

        return image, target

train_tf = T.Compose([
    T.RandomRatation(0),
    T.RandomRotation(10),
    T.RandomVerticalFlip(),
    T.RandomAffine(10, scale=(0.95, 1.05)),
])

train_ds = LiTS_Dataset("/content/Clean_LiTS/train_classes.csv", mode="train", transform = train_tf)
val_ds   = LiTS_Dataset("/content/Clean_LiTS/val_classes.csv",   mode="val")
test_ds  = LiTS_Dataset("/content/Clean_LiTS/test_classes.csv",  mode="test")


labels = []
for idx, row in train_ds.data.iterrows():
    liver_visible  = int(row["liver_visible"])
    lesion_visible = int(row["lesion_visible"])

    if liver_visible == 0:
        labels.append(0)      # keine Leber
    elif lesion_visible == 0:
        labels.append(1)      # Leber, kein Tumor
    else:
        labels.append(2)      # Leber + Tumor

labels = np.array(labels)

class_counts = np.bincount(labels)
print("Klassenhäufigkeiten:", class_counts)



class_weights = 1.0 / np.sqrt(class_counts)
class_weights = class_weights / class_weights.sum()  # normalisieren

print("Class Weights:", class_weights)


sample_weights = class_weights[labels]
sample_weights = torch.tensor(sample_weights, dtype=torch.float32)


sampler = WeightedRandomSampler(
    weights=sample_weights,
    num_samples=len(sample_weights),
    replacement=True
)
```

```python
import torch
import torch.nn as nn

device = "cuda" if torch.cuda.is_available() else "cpu"


class ImprovedCNN(nn.Module):
    def __init__(self):
        super().__init__()

        self.conv1 = nn.Sequential(
            nn.Conv2d(1, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.Conv2d(32, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU()
        )
        self.pool1 = nn.MaxPool2d(2)

        self.conv2 = nn.Sequential(
            nn.Conv2d(32, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.Conv2d(64, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU()
        )
        self.pool2 = nn.MaxPool2d(2)

        self.conv3 = nn.Sequential(
            nn.Conv2d(64, 128, 3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(),
            nn.Conv2d(128, 128, 3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU()
        )
        self.pool3 = nn.MaxPool2d(2)

        self.gap = nn.AdaptiveAvgPool2d((1,1))

        self.dropout = nn.Dropout(0.5)
        self.head_liver = nn.Linear(128, 1)
        self.head_tumor = nn.Linear(128, 1)

    def forward(self, x):
        x = self.pool1(self.conv1(x))
        x = self.pool2(self.conv2(x))
        x = self.pool3(self.conv3(x))

        x = self.gap(x)
        x = x.view(x.size(0), -1)

        x = self.dropout(x)

        out_liver = self.head_liver(x)
        out_tumor = self.head_tumor(x)

        return out_liver, out_tumor


model = ImprovedCNN().to(device)


class BCEFocalLoss(nn.Module):
    def __init__(self, alpha=0.25, gamma=2.0):
        super().__init__()
        self.alpha = alpha
        self.gamma = gamma
        self.bce = nn.BCEWithLogitsLoss(reduction="none")

    def forward(self, logits, targets):

        bce_loss = self.bce(logits, targets)
        pt = torch.exp(-bce_loss)
        loss = self.alpha * (1 - pt)**self.gamma * bce_loss
        return loss.mean()


focal_loss = BCEFocalLoss(alpha=0.25, gamma=2.0)




liver_pos = sum(train_ds.data["liver_visible"] == 1)
liver_neg = sum(train_ds.data["liver_visible"] == 0)
tumor_pos = sum(train_ds.data["lesion_visible"] == 1)
tumor_neg = sum(train_ds.data["lesion_visible"] == 0)

w_liver = liver_pos / (liver_pos + liver_neg)
w_tumor = tumor_pos / (tumor_pos + tumor_neg)

w_liver = 1.0 / w_liver
w_tumor = 1.0 / w_tumor


w_liver = torch.tensor(w_liver, dtype=torch.float32, device=device)
w_tumor = torch.tensor(w_tumor, dtype=torch.float32, device=device)

print("Liver weight:", w_liver.item())
print("Tumor weight:", w_tumor.item())


optimizer = torch.optim.Adam(model.parameters(), lr=1e-4)
```

```python

batch_size = 64

train_loader = DataLoader(
    dataset=train_ds,
    batch_size=batch_size,
    shuffle=True,
    num_workers=0,
    drop_last=True
)


val_loader = DataLoader(
    dataset=val_ds,
    batch_size=batch_size,
    shuffle=False,
    num_workers=0,
    drop_last=False
)

test_loader = DataLoader(
    dataset=test_ds,
    batch_size=batch_size,
    shuffle=False,
    num_workers=0,
    drop_last=False
)

print("train_loader ready:", len(train_loader))
print("val_loader ready:", len(val_loader))
print("test_loader ready:", len(test_loader))


num_epochs = 3

for epoch in range(num_epochs):


    model.train()

    for step, (data, targets) in enumerate(train_loader):

        data = data.to(device)
        targets = targets.to(device)

        optimizer.zero_grad()

        logit_liver, logit_tumor = model(data)

        t_liver = targets[:, 0].unsqueeze(1)
        t_tumor = targets[:, 1].unsqueeze(1)

        loss_liver = w_liver * focal_loss(logit_liver, t_liver)
        loss_tumor = w_tumor * focal_loss(logit_tumor, t_tumor)
        loss = loss_liver + loss_tumor

        loss.backward()
        optimizer.step()

        if step % 50 == 0:
            print(f"Epoch [{epoch+1}/{num_epochs}] "
                  f"Step [{step+1}/{len(train_loader)}] "
                  f"Loss: {loss.item():.4f}")


    if epoch % 1 == 0:
        model.eval()
        with torch.no_grad():

            losses = []
            batch_sizes = []

            all_preds = []
            all_targets = []

            for step, (data, targets) in enumerate(val_loader):
                data = data.to(device)
                targets = targets.to(device)

                logit_liver, logit_tumor = model(data)

                t_liver = targets[:, 0].unsqueeze(1)
                t_tumor = targets[:, 1].unsqueeze(1)

                loss_liver = w_liver * focal_loss(logit_liver, t_liver)
                loss_tumor = w_tumor * focal_loss(logit_tumor, t_tumor)
                loss = loss_liver + loss_tumor

                losses.append(loss.item())
                batch_sizes.append(data.size(0))

                liver_prob = torch.sigmoid(logit_liver)
                tumor_prob = torch.sigmoid(logit_tumor)

                liver_pred = (liver_prob >= 0.4).long()
                tumor_pred = (tumor_prob >= 0.4).long()


                for lv, tu in zip(liver_pred.cpu().numpy(), tumor_pred.cpu().numpy()):
                    lv = lv[0]
                    tu = tu[0]
                    if lv == 0:
                        all_preds.append(0)
                    elif tu == 0:
                        all_preds.append(1)
                    else:
                        all_preds.append(2)

                for lv, tu in zip(t_liver.cpu().numpy(), t_tumor.cpu().numpy()):
                    lv = lv[0]
                    tu = tu[0]
                    if lv == 0:
                        all_targets.append(0)
                    elif tu == 0:
                        all_targets.append(1)
                    else:
                        all_targets.append(2)

        avg_loss = sum([l * bs for l, bs in zip(losses, batch_sizes)]) / sum(batch_sizes)

        import numpy as np
        all_preds_np = np.array(all_preds)
        all_targets_np = np.array(all_targets)
        accuracy = np.mean(all_preds_np == all_targets_np)

        print(f"\nEpoch: {epoch+1}, Validation Loss: {avg_loss:.4f}, Accuracy: {accuracy:.4f}")

        class_correct = [0,0,0]
        class_total   = [0,0,0]

        for t, p in zip(all_targets_np, all_preds_np):
            class_total[t] += 1
            if t == p:
                class_correct[t] += 1

        print("Classwise Accuracy:")
        for i in range(3):
            if class_total[i] > 0:
                print(f"  Klasse {i}: {class_correct[i] / class_total[i]:.3f}")
            else:
                print(f"  Klasse {i}: ---")

        from sklearn.metrics import confusion_matrix
        import seaborn as sns
        import matplotlib.pyplot as plt

        cm = confusion_matrix(all_targets_np, all_preds_np)
        print("\nConfusion Matrix:\n", cm)

        plt.figure(figsize=(6,4))
        sns.heatmap(cm, annot=True, fmt="d", cmap="Blues")
        plt.title("Validation Confusion Matrix")
        plt.xlabel("Predicted")
        plt.ylabel("True")
        plt.show()

        model.train()
```

```python
model.eval()
with torch.no_grad():

    all_preds = []
    all_targets = []

    losses = []
    batch_sizes = []

    for data, targets in test_loader:

        data = data.to(device)
        targets = targets.to(device)

        logit_liver, logit_tumor = model(data)

        t_liver = targets[:, 0].unsqueeze(1)
        t_tumor = targets[:, 1].unsqueeze(1)

        loss_liver = w_liver * focal_loss(logit_liver, t_liver)
        loss_tumor = w_tumor * focal_loss(logit_tumor, t_tumor)
        loss = loss_liver + loss_tumor

        losses.append(loss.item())
        batch_sizes.append(data.size(0))

        liver_prob = torch.sigmoid(logit_liver)
        tumor_prob = torch.sigmoid(logit_tumor)

        liver_pred = (liver_prob >= 0.5).long()
        tumor_pred = (tumor_prob >= 0.5).long()

        for lv, tu in zip(liver_pred.cpu().numpy(), tumor_pred.cpu().numpy()):
            lv = lv[0]
            tu = tu[0]
            if lv == 0:
                all_preds.append(0)
            elif tu == 0:
                all_preds.append(1)
            else:
                all_preds.append(2)

        for lv, tu in zip(t_liver.cpu().numpy(), t_tumor.cpu().numpy()):
            lv = lv[0]
            tu = tu[0]
            if lv == 0:
                all_targets.append(0)
            elif tu == 0:
                all_targets.append(1)
            else:
                all_targets.append(2)




import numpy as np

all_preds_np = np.array(all_preds)
all_targets_np = np.array(all_targets)

avg_loss = sum([l * bs for l, bs in zip(losses, batch_sizes)]) / sum(batch_sizes)

accuracy = np.mean(all_preds_np == all_targets_np)

print(f"\nTEST RESULTS:")
print(f"Test Loss: {avg_loss:.4f}")
print(f"Test Accuracy: {accuracy:.4f}")

class_correct = [0,0,0]
class_total   = [0,0,0]

for t, p in zip(all_targets_np, all_preds_np):
    class_total[t] += 1
    if t == p:
        class_correct[t] += 1

print("\nClasswise Test Accuracy:")
for i in range(3):
    if class_total[i] > 0:
        print(f"  Klasse {i}: {class_correct[i] / class_total[i]:.3f}")
    else:
        print(f"  Klasse {i}: ---")

from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

cm = confusion_matrix(all_targets_np, all_preds_np)
print("\nConfusion Matrix:\n", cm)

plt.figure(figsize=(6,4))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues")
plt.title("Test Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("True")
plt.show()
```

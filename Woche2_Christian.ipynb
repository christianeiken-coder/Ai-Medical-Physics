### Chapter 1.2: "Loops, Iterables, and Conditions"

One of the most common patterns in programming is the loop. Loops go over some kind of iterable and perform some kind of action on every iteration. This seems very complicated until you realize that you do this in real life all the time. For example, when you go shopping for groceries, you perform a loop: Take your shopping list, check the next item, find it in the shop, put it in your bag, and repeat, until you have found every item.

To implement such a task in Python, we first require an iterable. An iterable is anything we can iterate on, i.e. something that can be split into similar parts: A list of groceries can be split into separate groceries and is therefore an iterable. My phonebook could be split into pairs of names and numbers and is therefore an iterable. A word like "banana" could be split into the individual characters "b", "a", "n", "a", "n", "a", and is therefore an iterable.

Secondly, we need a way to build a loop. For this, we have built-ins in python (see below).

````python
# Let's do our grocery list example again.

# This is our iterable, a List. We define it with square brackets. (We will learn more about lists later.)
grocery_list = ["milk", "eggs", "bread", "butter"]

# This is a for loop. It allows us to iterate over the items in our grocery list, and do something with each item.
for item in grocery_list:
    print(item)
````

The syntax always looks like this:
```
for X in Y:
    do something
```
X can be any name you want, but we want to use something appropriate as per usual:
`for item in grocery_list` makes sense, `for grocery in grocery_list` is also ok, etc.
Y has to be the name of the iterable.
"for" and "in" are fixed expressions in python, and we cannot change those.

You may have also noticed that the print command is indented (with a tab or 4 spaces) - Python uses what is called *significant whitespace*, meaning this empty area actually serves a purpose. Here, we know that we are still in the loop as long as the code we write is indented. Once we are no longer indented, we are out of the loop - forever. We will see this kind of indentation again later with functions. Remember: If it belongs together, it's indented together.

````python
# There is also a while loop, which allows us to do something as long as a certain condition is true:

i = 0
while i <= 5:
    print(i)
    i += 1  # This is the same as i = i + 1
````

#### Conditions

Sometimes, we don't know ahead of time what our inputs are going to be, and what we want to do or say in response. This also happens very often in a loop, because some iterables are very long.

For this, python implements simple conditions and ways to check for them. The syntax for these looks like this:

````python
secret_number = 7

for guess in [1, 4, 9, 7]:
    if guess < secret_number:
        print(f"{guess} is too low!")
    elif guess > secret_number:
        print(f"{guess} is too high!")
    else:
        print(f"{guess} is orrect!")
````

`if` gets followed by some kind of condition. Conditions are some kind of expression which can evaluate as either True or False. If they are True, we execute the indented code below the if-condition, otherwise we check the next condition (`elif`). If no conditions are left, we usually hit `else` and then just execute the code under that. If we have more situations that can happen, we can use as many `elif` conditions as we want.

````python
# We can also evaluate more than one condition at a time, using the 'and' and 'or' operators.

x = 5
y = 10
if x < 10 and y > 5:
    print("Both conditions are true.")
````

````python
if x < 10 or y < 5:
    print("At least one of the conditions is true.")
````

````python
# We can also negate a condition using the 'not' operator.

if not x > 10:
    print("x is not greater than 10.")
````

````python
# We can check if a variable is inside of something, such as a list or a string, using the 'in' operator.

if "milk" in grocery_list:
    print("Milk is on the grocery list.")
````

````python
# Finally, we can set conditions ahead of time, and then use them later in our code. This is called a conditional expression.

condition = x < 10

# We could do a bunch of other stuff here

if condition == True:
    print("x is less than 10.")
if condition: # This is the same as the previous if statement, but more concise.
    print("x is less than 10.")
````

#### Advanced loops with conditions

Let's see how if and else can be useful!

````python
# Let's solve the following task together. We have some numbers:
numbers = [7, 18, 1, 9, 65, 34]

# If the number is divisible by 3 or 5, we want to print it
# If the number is divisible by 13, we immediately want to stop the loop
# If the number is anything else, we ignore it

# Loops have some functionality which lets us skip an iteration or stop the loop altogether.
# Let's write some conditions to make use of these skips.

for n in numbers:
    if n % 13 == 0:
        print(f"We had to stop the loop, because I found {n}, which is divisible by 13. That's bad luck!")
        break # This stops the loop
    elif n % 3 == 0 or n % 5 == 0:
        print(f"I'm doing some code!")
    else:
        print(f"I don't care about {n}, because it is not divisible by 3, 5, or 13.")
        continue # This goes to the next step of the loop
    print(f"The current number is {n}. If you can read this, we did not run into break or continue.")

# Try to go over the output a few times so you understand what is printed when and why.
````

### Chapter 1.3: "Functions"

Next, we will look at functions. A function is a piece of code that you know you will use often, but which you do not want
to constantly rewrite and copy. The only thing you need to know about functions for now, is that they take *some* input and return
*some* output. We define functions with the 'def' keyword.

````python
# Any function has a signature which shows its name and what sort of inputs it accepts:

def myfunc(a, b):
    return a * b

# While python will not complain about this definition, it makes sense to include a variety of other information,
# such as so-called type hints and/or documentation.
# a: int means a is expected to be an integer, the -> int means the output is going to be an integer, too.

def myfunc(a: int, b: int) -> int:
    """
    This is a short description of the function's inputs and returns.
    a and b are expected to be integers, and the function will return an integer.
    """
    return a * b
````

````python
# We can use functions, and by default they will put their return value into the console,
# but we can also save the return value to a variable, and use it later.

myfunc(3, 3)  # This will print 9 to the console, but we won't be able to use it later.
result = myfunc(3, 3) # We will be able to use this later.
print(result)
````

````python
# Let's try putting in something that isn't a number, but rather text (called a string).
# You will notice that the result still sort of makes sense, but it is not what we expected.
# This is because Python is a dynamically typed language, and it will try to make sense of whatever you give it,
# instead of just throwing an error.

myfunc('3', 3)
````

````python
# While this flexibility is among the things that make python great, they can also be very confusing for beginners.

# When a function accepts is created, we can let others know which one we expect by setting type hints.
# When it accepts multiple data types, we can let others know by relaxing the type hints:

def myfunc(a: int | float, b: int | float) -> int | float:
    """
    This function returns the product of two numbers a and b.
    The inputs and outputs could be int or float
    """
    return a * b

# Note that type hints are just that - hints. They are not enforced by Python itself.
````

````python
# Not all functions have inputs or return something - this is a perfectly valid function ...

def no_inputs() -> int:
    some_var = 5
    return some_var

# ... but so is this ...

def no_returns(some_arg: str, some_other_arg: str) -> None:
    print(f'The inputs to this function call were {some_arg} and {some_other_arg}')

# ... and this (even if its completely useless):
def do_nothing() -> None:
    pass
````

#### Decorators

What Are Decorators in Python?

A decorator in Python is a higher-order function that modifies or extends the behavior of another function without modifying its actual code. Decorators are commonly used for logging, timing execution, enforcing access control, memoization (caching), and more.

How Do Decorators Work?

A decorator is a function that takes another function as an argument, wraps it in another function (usually called wrapper), and returns the wrapped function.


````python
import time

def timing_decorator(func):
    """Decorator that times the execution of a function."""
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)
        end_time = time.time()
        elapsed_time = end_time - start_time
        print(f"Function {func.__name__} took {elapsed_time:.6f} seconds to execute.")
        return result
    return wrapper

# Example usage
@timing_decorator
def example_function():
    time.sleep(1)
    return "Finished"

# Run the function
example_function()
````

#### Chapter 1.5.3: "Tuples"

````python
# Tuples are a bit like lists. However, they have one key difference: Once you make a tuple,
# you can't change it anymore. They are 'immutable'.
# This may sound strange and plainly less useful than lists, but it can sometimes be faster
# to use tuples over lists. There is also another feature you will find out about in a bit.

# You can declare a tuple like this:
my_tuple = (1, 2)

# Just like lists and dicts, you can put basically anything into a tuple:
another_tuple = ("one", 2.0, 3, "4")
````

````python
# You can look at parts of your tuple just like with lists:
print(another_tuple[0])
print(another_tuple[0:3])
````

````python
# You can check if something is in the tuple:
if "4" in another_tuple:
    print("Yup, '4' is in my tuple.")
````

````python
# Now, what are tuples actually good for?
# Python comes with a neat feature called tuple unpacking. You can see this feature in action below:

my_tuple = (1, 2, 3)
a, b, c = my_tuple
print(a)

# Python just saw that I wanted to declare three variables. On the other side of the '=' was a tuple
# with three entries, and Python assigned the first entry to my first variable, second to second, etc.
````

````python
# Why is this useful?
# Let's say I have a function that calculates several things for me. I want to get back all the
# different outputs, and I will be using them for separate things. Tuple unpacking will take the
# output from the function and try to 'unpack' it into however many variables I suggest:

def my_function(a: int, b: int):
    c = a + b
    d = a - b
    return c, d

e = 3
f = 2

# Here I suggest g and h as variables, and tuple unpacking will expect my function to give back
# a tuple of length two.
g, h = my_function(a = e, b = f)
print(g)
print(h)
````

### Chapter 1.6: "List Comprehensions"

Python contains a nifty little feature called List Comprehension. This allows you to write a piece of code that
would otherwise be a little unwieldy in a single short expression.

````python
# The following two segments of code are equivalent and both create a list containing
# the squares of the digits from 0 to 9:

# Method 1
squared_digits = []
for x in range(10):
    y = x ** 2
    squared_digits.append(y)
print(squared_digits)

# Method 2
squared_digits = [x ** 2 for x in range(10)]
print(squared_digits)

# We have essentially just executed an entire loop inside of the square brackets of
# our list, at the moment of its creation.
````

````python
# This tool also works on other data types, such as tuples or dictionaries:
squared_digits = {x: x**2 for x in range(10)}
print(squared_digits)
````

````python
# It can even handle more complex instructions, such as conditionals.
# Let's make a list that contains only digits that are even.
only_even_digits = [x for x in range(10) if x % 2 == 0]
print(only_even_digits)
````

````python
# Let's make one that containts even and odd digits, but even digits are squared.
square_even_digits = [x ** 2 if x % 2 == 0 else x for x in range(10)]
print(square_even_digits)

# List comprehensions are an extremely powerful tool which is worth practicing a little.
````

````python
# Let's make a list of primes up to 100:


# First, we make an is_prime function - we've done this earlier:

def is_prime(dividend: int):
    # If the dividend is smaller than 2, it can't be prime
    if dividend < 2:
        return False
    # If the dividend is two, it is a prime.
    elif dividend == 2:
        return True
    else:
        # Try all divisors starting at 2 and up to the square root of the dividend
        for divisor in range(2, max(3, int(np.sqrt(dividend)+1))):
            if dividend % divisor == 0:
                return False
            else:
                pass
                # 'pass' means 'do nothing' - instead of writing
                # else: pass, you can also simply write nothing
        # If we have gotten past the loop without finding a divisor
        # (and thus returning it), the number is a prime.
        return True

# Solution with function and no list-comp:
primes = []
for x in range(1, 101):
    if is_prime(x):
        primes.append(x)
print(primes)
````

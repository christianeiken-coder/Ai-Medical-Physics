### Question 23 — Factorial

**3 pts** · **Type:** Programming · **Difficulty:** easy · **Source:** P8 · *easy implementation*

Write `factorial(n)` returning `n!` for an integer `n >= 0`. By definition, `0! == 1`. You may use a loop or recursion. Do **not** use `math.factorial`.

````python
def factorial(n):
    """Return n! for n >= 0."""
    if n == 0:
      return 1
    if n == 1:
      return 1
    if n == 2:
      return 2

    for i in range(2, n+1):
      F = factorial(n-1)*n
      return F

````

````python
print(factorial(0))  # expected 1
print(factorial(4))  # expected 24
print(factorial(6))  # expected 720
````

### Question 24 — Count values in a closed interval

**3 pts** · **Type:** Programming · **Difficulty:** easy · **Source:** B22 · *easy implementation*

Write `count_in_range(values, low, high)` returning how many values `v` satisfy `low <= v <= high`. Both endpoints are included. Return `0` for an empty list.

````python
def count_in_range(values, low, high):
    count = 0
    for v in values:
        if low <= v <= high:
            count += 1
    return count

def count_in_range(values, low, high):
    return sum(1 for v in values if low <= v <= high)
````

````python
print(count_in_range([0, 5, 10], 0, 10))  # expected 3
print(count_in_range([], 0, 10))           # expected 0
print(count_in_range([-1, 0, 1], 0, 1))    # expected 2
````

### Question 25 — One-pass summary

**4 pts** · **Type:** Programming · **Difficulty:** hard · **Source:** A21 · *difficult implementation*

A sensor produces a list of measurements. Write `summarize(values, limit)` returning a tuple `(mean, maximum, count_high)`:

- `mean` is the arithmetic mean as a float.
- `maximum` is the largest value.
- `count_high` is the number of values **strictly greater** than `limit`.
- For an empty list, return `(0.0, None, 0)`.
- Use one loop. Do **not** use `sum`, `max`, or `statistics`.

````python
def summarize(values, limit):
    if not values:
        return (0.0, None, 0)

    total = 0
    maximum = None
    count_high = 0

    for v in values:
        total += v
        if maximum is None or v > maximum:
            maximum = v
        if v > limit:
            count_high += 1

    return (total / len(values), maximum, count_high)
````

````python
print(summarize([1, 2, 3, 4], 3))  # expected (2.5, 4, 1)
print(summarize([], 0))             # expected (0.0, None, 0)
````

### Question 27 — Bonus: Collatz step count

**4 pts** · **Type:** Programming · **Difficulty:** hard · **Source:** P7 · *bonus*

The Collatz sequence for a positive integer `n` is built by repeatedly applying

- `T(n) = 3n + 1` if `n` is odd;
- `T(n) = n // 2` if `n` is even.

Each application counts as **one step**. The sequence is complete when it reaches `1`. For `n = 1`, the step count is `0`.

Write `collatz_steps(n)` returning the number of steps needed to reach `1`. Assume `n >= 1`.

This is a bonus question. No negative marking; leave it blank if you run out of time.

````python
def collatz_steps(n):
    steps = 0
    while n != 1:
        if n % 2 == 1:
            n = 3 * n + 1
        else:
            n = n // 2
        steps += 1
    return steps
````

````python
print(collatz_steps(1))   # expected 0
print(collatz_steps(6))   # expected 8
print(collatz_steps(27))  # expected 111
````
